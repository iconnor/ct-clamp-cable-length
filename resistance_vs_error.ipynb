# CT Clamp Cable Length - Resistance vs Error Analysis

This notebook analyzes the relationship between cable resistance and measurement error for CT clamps at different power levels (three-phase systems).

```python
import matplotlib.pyplot as plt

# Power levels (three phase) in watts
power_w = [10, 100, 1000, 5000, 10000]

# Approximate relative error (%) for each extra burden
error_1ohm = [300, 200, 12, 5, 2.5]
error_2ohm = [600, 400, 24, 10, 5]
error_3ohm = [900, 600, 36, 15, 7.5]

plt.figure(figsize=(8, 5))

# Plot error vs power for each resistance
plt.plot(power_w, error_1ohm, marker='o', label='1Ω Extra Burden', linewidth=2)
plt.plot(power_w, error_2ohm, marker='s', label='2Ω Extra Burden', linewidth=2)
plt.plot(power_w, error_3ohm, marker='^', label='3Ω Extra Burden', linewidth=2)

# Use logarithmic scale for both axes as data spans multiple orders of magnitude
plt.xscale('log')
plt.yscale('log')

# Labels and title
plt.xlabel('Power (W)', fontsize=12)
plt.ylabel('Relative Error (%)', fontsize=12)
plt.title('CT Clamp Measurement Error vs Power Level\nfor Different Cable Resistances', fontsize=14, fontweight='bold')

# Grid for better readability
plt.grid(True, which='both', linestyle='--', alpha=0.6)

# Legend
plt.legend(loc='best', fontsize=10)

# Tight layout for better spacing
plt.tight_layout()

# Display the plot
plt.show()
```

## Key Observations

1. **Error decreases with power**: Higher power levels result in lower relative errors
2. **Linear relationship with resistance**: Doubling the cable resistance approximately doubles the error
3. **Critical at low power**: At 10W, even 1Ω of extra burden results in 300% error
4. **Acceptable at high power**: At 10kW, even 3Ω of extra burden results in only 7.5% error

## Recommendations

- For low-power applications (< 100W), minimize cable length and resistance
- For high-power applications (> 1kW), cable resistance has less impact
- Consider using thicker cables (lower resistance) for long runs
